```python
# Create a new column as a result of a mathematical operation (e.g., 'age' and 'tumor_size')

df['Age_Tumor']=df['Age'] + df['TumorSize']
df[['Age_Tumor','Age','TumorSize']]
```

```python
# Rename 'tumor_size' to 'size_of_tumor'

df.rename(columns={'TumorSize':'Size_of_tumor'},inplace=True)
df.columns
```

```python
# Create a new DataFrame with 'age' and 'tumor_size', where 'tumor_size' > 5

df_new=df[df['Size_of_tumor']>5] [['Age','Size_of_tumor']]
df_new
```
